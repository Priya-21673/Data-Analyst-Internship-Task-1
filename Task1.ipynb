```python
# ============================================================
# DATA ANALYST INTERNSHIP - TASK 1
# DATA CLEANING AND PREPROCESSING
# ============================================================

# -------------------------------
# 1. IMPORT REQUIRED LIBRARIES
# -------------------------------

import pandas as pd
import numpy as np


# -------------------------------
# 2. LOAD THE DATASET
# -------------------------------

# Change the file name/path if necessary
# Ensure 'marketing_campaign.csv' is uploaded to your Colab environment
df = pd.read_csv("/content/drive/MyDrive/marketing_campaign.csv", sep="\t")

print("Dataset loaded successfully!")


# -------------------------------
# 3. VIEW THE ORIGINAL DATASET
# -------------------------------

print("\nFIRST 5 ROWS:")
print(df.head())

print("\nLAST 5 ROWS:")
print(df.tail())

print("\nDATASET SHAPE:")
print(df.shape)

print("\nCOLUMN NAMES:")
print(df.columns)

print("\nDATASET INFORMATION:")
df.info()


# -------------------------------
# 4. CHECK MISSING VALUES
# -------------------------------

print("\nMISSING VALUES BEFORE CLEANING:")
print(df.isnull().sum())

print("\nTOTAL MISSING VALUES:")
print(df.isnull().sum().sum())


# -------------------------------
# 5. HANDLE MISSING VALUES
# -------------------------------

# Check if Income column exists
if "Income" in df.columns:

    # Fill missing income values using median
    df["Income"] = df["Income"].fillna(df["Income"].median())

    print("\nMissing values in Income handled using median.")


# -------------------------------
# 6. CHECK DUPLICATE RECORDS
# -------------------------------

print("\nNUMBER OF DUPLICATE ROWS:")
print(df.duplicated().sum())


# -------------------------------
# 7. REMOVE DUPLICATE RECORDS
# -------------------------------

df = df.drop_duplicates()

print("\nDuplicate rows removed.")


# -------------------------------
# 8. CLEAN COLUMN NAMES
# -------------------------------

df.columns = (
    df.columns
    .str.strip()                    # Remove extra spaces
    .str.lower()                    # Convert to lowercase
    .str.replace(" ", "_")          # Replace spaces with underscores
)

print("\nCLEANED COLUMN NAMES:")
print(df.columns)


# -------------------------------
# 9. STANDARDIZE TEXT DATA
# -------------------------------

# List of text columns to clean
text_columns = ["education", "marital_status"]

for column in text_columns:

    if column in df.columns:

        # Remove extra spaces
        # Convert text to title case

        df[column] = (
            df[column]
            .astype(str)
            .str.strip()
            .str.title()
        )

print("\nTEXT VALUES STANDARDIZED.")


# -------------------------------
# 10. CHECK UNIQUE VALUES
# -------------------------------

if "education" in df.columns:
    print("\nUNIQUE EDUCATION VALUES:")
    print(df["education"].unique())

if "marital_status" in df.columns:
    print("\nUNIQUE MARITAL STATUS VALUES:")
    print(df["marital_status"].unique())


# -------------------------------
# 11. CONVERT DATE FORMAT
# -------------------------------

if "dt_customer" in df.columns:

    df["dt_customer"] = pd.to_datetime(
        df["dt_customer"],
        errors="coerce"
    )

    print("\nDATE COLUMN CONVERTED TO DATETIME.")


# -------------------------------
# 12. CHECK DATA TYPES
# -------------------------------

print("\nDATA TYPES:")
print(df.dtypes)


# -------------------------------
# 13. FIX DATA TYPES
# -------------------------------

# Convert selected columns to integer
integer_columns = [
    "year_birth",
    "kidhome",
    "teenhome",
    "recency",
    "mntwines",
    "mntfruits",
    "mntmeatproducts",
    "mntfishproducts",
    "mntsweetproducts",
    "mntgoldprods"
]

for column in integer_columns:

    if column in df.columns:

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )


print("\nDATA TYPES CHECKED AND CORRECTED.")


# -------------------------------
# 14. CREATE AGE COLUMN
# -------------------------------

if "year_birth" in df.columns:

    current_year = 2026

    df["age"] = current_year - df["year_birth"]

    print("\nAGE COLUMN CREATED.")


# -------------------------------
# 15. CHECK AGE STATISTICS
# -------------------------------

if "age" in df.columns:

    print("\nAGE STATISTICS:")
    print(df["age"].describe())


# -------------------------------
# 16. REMOVE UNREALISTIC AGE VALUES
# -------------------------------

if "age" in df.columns:

    original_rows = df.shape[0]

    # Keep only reasonable age values
    df = df[
        (df["age"] >= 18) &
        (df["age"] <= 100)
    ]

    final_rows = df.shape[0]

    print("\nRows before age cleaning:", original_rows)
    print("Rows after age cleaning:", final_rows)


# -------------------------------
# 17. CHECK FOR OUTLIERS USING IQR
# -------------------------------

if "income" in df.columns:

    Q1 = df["income"].quantile(0.25)
    Q3 = df["income"].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df["income"] < lower_bound) |
        (df["income"] > upper_bound)
    ]

    print("\nINCOME OUTLIERS:")
    print(outliers[["income"]])

    print("\nNUMBER OF OUTLIERS:")
    print(outliers.shape[0])


# -------------------------------
# 18. FINAL MISSING VALUE CHECK
# -------------------------------

print("\n========================================")
print("FINAL DATA QUALITY CHECK")
print("========================================")

print("\nMISSING VALUES AFTER CLEANING:")
print(df.isnull().sum())


# -------------------------------
# 19. FINAL DUPLICATE CHECK
# -------------------------------

print("\nDUPLICATE ROWS AFTER CLEANING:")
print(df.duplicated().sum())


# -------------------------------
# 20. FINAL DATA TYPES
# -------------------------------

print("\nFINAL DATA TYPES:")
print(df.dtypes)


# -------------------------------
# 21. STATISTICAL SUMMARY
# -------------------------------

print("\nSTATISTICAL SUMMARY:")
print(df.describe())


# -------------------------------
# 22. FINAL DATASET SHAPE
# -------------------------------

print("\nFINAL DATASET SHAPE:")
print(df.shape)


# -------------------------------
# 23. DISPLAY CLEANED DATA
# -------------------------------

print("\nCLEANED DATASET - FIRST 5 ROWS:")
print(df.head())


# -------------------------------
# 24. SAVE CLEANED DATASET
# -------------------------------

df.to_csv(
    "cleaned_customer_personality_data.csv",
    index=False
)

print("\n========================================")
print("DATA CLEANING COMPLETED SUCCESSFULLY!")
print("Cleaned dataset saved as:")
print("cleaned_customer_personality_data.csv")
print("========================================")
```
